```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd

data=np.array(pd.read_csv('olympic100m.csv', header=None))
n=len(data)
np.random.seed(234)
# np.random.shuffle(data)
train, test = np.split(data, [19])

test_x = np.array(test[:, 0])
test_y = np.array(test[:, 1])

train_x = np.array(train[:, 0])
train_y = np.array(train[:, 1])
```

```python
# 下週考五等分

# 1st-order
import statsmodels.api as sm

# modeling
x = sm.add_constant(train_x)
model = sm.OLS(train_y ,x)
result = model.fit()
# print(result.summary())

# predict
test_x1 = sm.add_constant(test_x)
prediction = result.predict(test_x1)

# error
err = 0
for i in range(len(prediction)):
    err += (test_y[i] - prediction[i])**2
    # print(test_y[i], prediction[i])

print(np.log10(err))
print(err)
```

```python
# 6th-order
import statsmodels.api as sm

# modeling 
train_x6 = np.column_stack([train_x**i for i in range(1, 7)])
train_x6 = sm.add_constant(train_x6)
model = sm.OLS(train_y ,train_x6)
result = model.fit()
print(result.summary())

# predict
test_x6 = np.column_stack([test_x**i for i in range(1, 7)])
test_x6 = sm.add_constant(test_x6)
prediction = result.predict(test_x6)

# error
err = 0
for i in range(len(prediction)):
    err += (test_y[i] - prediction[i])**2
    # print(test_y[i], prediction[i])

print(np.log10(err))
print(err)
```
