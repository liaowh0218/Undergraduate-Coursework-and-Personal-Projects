```python
# 課本 feature 7.9
from scipy.stats import beta
from matplotlib import pyplot as plt
import numpy as np

n=50000
np.random.seed(234)
intermediate=[]
trajectory=[]
cur = [0.7,0.3]
for i in range(n):
    cur[0]=beta.rvs(6+2, 8-6+2)
    intermediate.append(cur.copy()) 
    cur[1]=beta.rvs(2+2, 7-2+2)
    intermediate.append(cur.copy())
    trajectory.append(cur.copy())

trajectory = np.array(trajectory)
intermediate = np.array(intermediate)
plt.plot(trajectory[:1000,0], trajectory[:1000,1])
```

```python
def std_err(a):
    ''' 'a' is a list, each data in list contains coordinates of (x, y) '''
    dist = []
    for i in range(len(a)-1):
        dis = a[i][0] - a[i][1]
        dist.append(dis)

    se = np.std(dist, ddof=1)     #se is std of sampling distribution
    print(se)

std_err(intermediate)
std_err(trajectory)
```

```python
trajectory = trajectory.tolist()
intermediate = intermediate.tolist()

mid_point = [x for x in intermediate if x not in trajectory]
std_err(mid_point)
```
