```python
# 投2個骰子24次出現至少1次[6, 6]的機率
import random as rd

out = []           #sample space
for i in range(6):
    for j in range(6):
        out.append([i+1, j+1])

count = 0
mock = 100000
rd.seed(234)
for i in range(mock):
    totalout = []
    for i in range(24): # sampling 24 times
        totalout.append(out[rd.randint(0, len(out)-1)])
    if [6, 6] in totalout:
        count += 1

prob = count/mock
print(prob)
```

```python
# 三門遊戲，換選擇之勝率
import random as rd
rd.seed(234)

switchWin = 0
switchLoss = 0
car = 1
mock = 100000
for i in range(mock):
    first = rd.randint(1, 3)
    if first != car:
        open = 6 - car - first
    else:
        open = rd.randint(2, 3)

    switch = 6 - open - first
    if switch == car:
        switchWin += 1
    else:
        switchLoss += 1

WinRate = switchWin/(mock)
print(WinRate)
```

```python
# 四門遊戲，換選擇之勝率
import random as rd
rd.seed(234)

switchWin = 0
switchLoss = 0
car = 1
mock = 100000
for i in range(mock):
    l = [1,2,3,4]
    first = rd.randint(1, 4)
    if first != car:
        l.remove(car)
        l.remove(first)
        open = rd.choice(l)
        l.extend([car, first])
        l.remove(open)
    else:
        l.remove(car)
        open = rd.choice(l)
        l.append(car)
        l.remove(open)

    l.remove(first)
    switch = rd.choice(l)
    if switch == car:
        switchWin += 1
    else:
        switchLoss += 1

WinRate = switchWin/mock
print(WinRate)
```
